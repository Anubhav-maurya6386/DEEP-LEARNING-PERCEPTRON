### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

dx = x2 - x1
dy = y2 - y1

dx_square = dx * dx
dy_square = dy * dy

sum_value = dx_square + dy_square

distance = sum_value ** 0.5

print("Euclidean Distance is:", distance)
````

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
# Write your code here
principal = float(input("Enter principal amount: "))
rate = float(input("Enter rate of intreast: "))
time = float(input("Enter the time period: "))
simple_interest = (principal * rate * time)/100;
print("Simple interest is:",simple_interest)
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads = int(input("Enter number of heads: "))
legs = int(input("Enter number of legs:"))
extra = legs - 2*heads
dog = extra//2
chicken = heads - dog
print("number of dogs is:",dog)
print("number of chicken is:",chicken)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
first_term = int(input("Enter first term of AP:"))
second_term = int(input("Enter second term of AP:"))
n = int(input("Enter the value of n:"))
d = second_term - first_term
nth_term = first_term + (n-1) * d
print("Nth term of AP is:",nth_term)
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
# Write your code here
num1 = int(input("Enter numerator of first fraction:"))
den1 = int(input("Enter denominator of first fraction:"))
num2 = int(input("Enter numerator of second fraction:"))
den2 = int(input("Enter denominator of second fraction:"))
if den1 == den2:
  sum_num = num1 + num2
  sum_den = den2;
  print("Sum of fraction is:",sum_num,"/",sum_den)
else:
  sum_num = num1 * den2 + num2 * den1
  sum_den = den1 * den2
  print("Sum of fraction is:",sum_num,"/",sum_den)
````
